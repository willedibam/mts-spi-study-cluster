# Proof figure, 2026-11-26

The proof-of-concept figure for the paper.
- **Panel a** is one t-SNE of nine generator classes. One silo per family surrounds it, each variant tiled with carpets of example recordings whose height encodes M.
- **Panel b** shows four fixed class contrasts.

Data are the 2026-11-26 bank: the recordings of the 2026-10-14 corpus recomputed with the released pyspi 3.0.0rc2 (bundled `benchmarked_p90`, 207 SPIs), with ordered z (`unified_ordered_v3`). [proof_figure_261029](proof_figure_261029.ipynb) is the same figure on the earlier 283-SPI r9 bank.

Every setting is frozen in `research/proof/p90/configs/analysis/proof-figure-261126.yaml`: classes, colours, t-SNE, KDE islands, contrasts and panel-a layout. The code is `research/proof/p90/proof_figure.py`.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
ROOT = Path.cwd().resolve()
while not (ROOT / 'src').is_dir() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
from research.proof.p90.refresh_proof_figures import load_bank
from research.proof.p90.proof_figure import (load_config, fit_embedding, select_contrasts, choose_layout,
                                             choose_strips, silo_carpets, plot_proof_figure)
FIGURE = load_config(ROOT / 'research/proof/p90/configs/analysis/proof-figure-261126.yaml')
OUT = ROOT / 'results/proof/figure-261126'
bank = load_bank(ROOT / 'data/proof/refresh-261126/features.npz', expected_spis=207, contract='unified_ordered_v3')
```

## Embedding and checks

Preprocessing and PCA50 are fitted on the nine displayed classes only, as in the 261014 protocol. t-SNE then uses openTSNE with Student-t degrees of freedom 5. The lighter kernel tails spread points within islands so that M and T stay legible ([Kobak et al. 2020](https://arxiv.org/abs/1902.05804)).

Check that the classes are already separated before t-SNE: a 5-nearest-neighbour classifier on PCA50, holding out each M/T cell in turn. Preprocessing and PCA see every cell, so this is not a fully held-out test.

```python
embedding = fit_embedding(bank, OUT, FIGURE)
print(f"{embedding['audit']['datasets']} recordings, {embedding['audit']['features']:,} coordinates; "
      f"PCA50 5NN accuracy, each M/T cell held out: {embedding['audit']['pca50_knn5_cell_held_out_accuracy']:.3f}")
contrasts = select_contrasts(bank, OUT, FIGURE)
contrasts[['class_a', 'class_b', 'spi_a', 'spi_b', 'training_auc_oriented', 'display_auc_fixed_orientation']].round(3)
```

## Example recordings

Each variant's strip is tiled by a frozen layout (`panel_a.silos.layouts`, from `choose_strips` below), and the tiles are filled on every run.
- **Height encodes M.** Carpets are `unit_in` per 8 channels, less a gap, so two M carpets stack to exactly one 2M. Side strips hold 48 channels. The Wave strip holds 64; its unit stretches about 1.12× to fill the height left after aligning the side columns with the embedding.
- **Width is a time window,** solved per layout so each strip fills edge to edge, at 238 samples per inch. Intermittency shows every 12th sample, and Kuramoto shows the whole record decimated to the tile. A record too short for its tile is stretched, not repeated.
- **Which recordings.** Tiles take recordings with their M. Wider tiles take longer recordings (T by width tercile), and instances run 6–9, then 0–5, without repeats within a strip. The carpets are not tied to t-SNE points, so no points are ringed.
- Recordings are regenerated from the frozen generation config and scaled robustly per channel, using icefire limited to the 99th percentile of |x|.

## Layout

t-SNE axes carry no meaning, so the map is rotated or reflected once. `choose_layout` searched silo sides and orders, the variant order within each silo, and orientation. It favours short leaders that neither cross each other nor run over other islands, with a free corner for the inset key. The Wave silo is held above the embedding (`top=('Wave equation',)`, the user's draw.io arrangement) and the other families are searched left/right. The result is frozen in the YAML (`panel_a`). Set `SEARCH = True` to rerun it (about 20 s).

Leaders are routed per variant at draw time. Each starts at the point of its section's inner edge that gives a short leader clear of the other silos, the key, earlier leaders and other islands. If no straight path clears, it elbows down the gap beside the embedding.

`STRIPS = True` reranks the strip layouts (`carpet_strips.rank_layouts`) and writes a candidate sheet. Side variants cycle through the three best distinct layouts, mirrored on alternate rounds. Paste the result into `panel_a.silos.layouts`, or edit layout strings by hand: `|` columns, `,` stacked levels, `+` side by side, `/` bands.


```python
SEARCH, STRIPS = False, False
if SEARCH:
    print(choose_layout(embedding['tsne'], bank['y'][embedding['indices']], FIGURE, top=('Wave equation',)))
else:
    print({k: FIGURE['panel_a'][k] for k in ['columns', 'variant_order', 'rotation_deg', 'reflect_x']})
if STRIPS:
    for name, layout in choose_strips(FIGURE, preview=OUT/'strip-candidates.png').items(): print(f"{name}: '{layout}'")
carpets = silo_carpets(FIGURE)
```

## Caption notes

- **Colour** gives the class; a family's variants take neighbouring hues. Marker area ∝ M; shape gives T.
- **Islands** are tinted 95% highest-density regions of a Gaussian KDE, with a thin contour. Island areas, gaps, positions and contour shapes carry no meaning, and the map is rotated.
- **Carpets** are example recordings of each variant. Height ∝ M (the Wave silo ~1.12× taller per channel); width is a time window, not T. Intermittency is strobed every 12th sample (SPIs used every sample) and Kuramoto shows whole records.
- **Silo glyphs** show each variant's coupling, with line width ∝ weight: self-loops for a node's own past and ring edges per neighbour (VAR, CML; near-zero edges dotted). Kuramoto's all-to-all coupling is drawn through a mean-field hub, dashed because it is repulsive. Noise has no coupling; the wave equation is a periodic ring. Generator equations and parameters belong in the SI table.
- **Panel b.** Each contrast coordinate was chosen for mechanistic readability among coordinates with training AUC ≥ .995 in every M/T cell (instances 0–5). The rainclouds show instances 6–9. These are illustrations, not unique or confirmatory axes. Readings are in the YAML (`contrasts[].reading`).
